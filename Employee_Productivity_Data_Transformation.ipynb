## Q7 Build Preprocessing PipelineCreate a pipeline that applies encoding to categorical columns and scaling to numerical columns,using `ColumnTransformer` and `Pipeline`.

```python
from sklearn.compose import ColumnTransformerfrom sklearn.pipeline import Pipelinefrom sklearn.preprocessing import OneHotEncoder, StandardScaler# Reload a fresh copy of the raw data so the pipeline handles encoding/scaling itselfraw_df = pd.read_csv("employee_productivity_dataset.csv")raw_df["Age"] = raw_df["Age"].fillna(raw_df["Age"].median())raw_df["Salary"] = raw_df["Salary"].fillna(raw_df["Salary"].mean())raw_df["Hours_Worked_Per_Week"] = raw_df["Hours_Worked_Per_Week"].fillna(raw_df["Hours_Worked_Per_Week"].median())raw_df["Performance_Score"] = raw_df["Performance_Score"].fillna(raw_df["Performance_Score"].mean())categorical_cols = ["Gender", "Department", "Work_Mode", "Location"]numerical_cols = ["Age", "Hours_Worked_Per_Week", "Projects_Completed", "Performance_Score", "Salary"]preprocessor = ColumnTransformer(transformers=[    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),    ("num", StandardScaler(), numerical_cols)])pipeline = Pipeline(steps=[    ("preprocessor", preprocessor)])pipeline
```

## Q8 Apply PipelineApply the pipeline on the dataset and display the transformed dataset and its shape.

```python
transformed_array = pipeline.fit_transform(raw_df[categorical_cols + numerical_cols])feature_names = pipeline.named_steps["preprocessor"].get_feature_names_out()transformed_df = pd.DataFrame(transformed_array, columns=feature_names)print("Shape of final transformed dataset:", transformed_df.shape)transformed_df.head()
```

## Q9 Conceptual Question -- Why is scaling important in Python?Scaling brings all numerical features onto a comparable range or distribution. Many ML algorithms(KNN, SVM, logistic/linear regression with gradient descent, PCA, neural networks) are distance-basedor gradient-based, so a feature like Salary (values in tens of thousands) would dominate a featurelike Age (values in tens) purely because of its larger scale, not because it's more important.Scaling ensures each feature contributes proportionally, speeds up model convergence, and avoidsbiased or misleading model weights.

## Q10 Conceptual Question -- Why do we convert categorical data into numerical form?Most machine learning algorithms are mathematical models that operate on numbers -- they cannotdirectly interpret text labels like "Male"/"Female" or "IT"/"HR". Converting categorical data intonumerical form (via Label Encoding or One-Hot Encoding) allows these algorithms to process theinformation, calculate distances, gradients, or probabilities, and learn patterns from categoricalfeatures just like they do from numerical ones.
